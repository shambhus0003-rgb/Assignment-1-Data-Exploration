```python
# 1. String Concatenation
# Storeing the initial greeting word in str1
str1 = "Hello"

#  enter the name
str2 = input("Enter your Name: ")

# Join str1, a space (" "), and str2 using '+'
greeting = str1 + " " + str2
print(greeting)

# Define the welcome message to append at the end
str3 = ", welcome to Python programming"

# Concatenate str3 onto the existing greeting string and print the full message
full_greeting = greeting + str3
print(full_greeting)
```

```python
#2. String Slicing and Indexing
# a. Index 0 accesses the very first character of the string
print("a. First character:", full_greeting[0])

# b. Index -1 accesses the very last character from the end of the string
print("b. Last character:", full_greeting[-1])

# c. [:5] starts at index 0 and stops before index 5, giving the first 5 characters
print("c. First 5 characters:", full_greeting[:5])

# d. [-11:] starts 11 characters from the end and goes to the end of the string
print("d. Last 11 characters:", full_greeting[-11:])

# e. [::-1] steps backward (-1) across the entire string to reverse it
print("e. Reversed string:", full_greeting[::-1])

# f. Negative slicing [-18:-12] extracts "Python" reliably regardless of name length,
# because ", welcome to Python programming" has a fixed length at the end of the string
print("f. Word 'Python':", full_greeting[-18:-12])
```

```python
# 3. String Methods
strM = "Python beginner tutorial"

# a. Convert the sentence to uppercase
print("a:", strM.upper())

# b. Convert the sentence to lowercase
print("b:", strM.lower())

# c. Use capitalize and return the sentence to the original input form
print("c:", strM.capitalize())

# d. Count the total number of occurrences of character 't' in the string
print("d:", strM.count('t'))

# e. Replace all occurrences of "Python" with "Machine Learning"
print("e:", strM.replace("Python", "Machine Learning"))
```

```python
# 4.Tuples (Creation, Modification, and Access)
# Create the two tuples
t1 = (10, 20, 30)
t2 = (40, 50, 60)

# a. Concatenate the two tuples and store it in "t_combine"
t_combine = t1 + t2
print("a:", t_combine)

# b. Repeat the elements of "t_combine" 3 times
t_repeat = t_combine * 3
print("b:", t_repeat)

# c. Access the 3rd element from "t_combine"
print("c:", t_combine[2])

# d. Access the first three elements from "t_combine"
print("d:", t_combine[:3])

# e. Access the last three elements from "t_combine"
print("e:", t_combine[-3:])
```
